**Paso 4:** Defina el modelo del sistema y los objetivos de la simulación

Tenemos las 5 v.a.´s las cuales tienen una distribución trinagular, por eso usamos r.triangular. Dónde 190 es el extremo izquierdo, 230 el derecho y 210 el medio (seguramente la moda). Así, sumamos los 5 datos y creamos una función en la cuál nos arroja el no. de datos que le estamos asignando a n (ya sumando las 5 v.a.´s).

```python
def S(n):
  tinas = []
  x = 0
  for a in range(n):
    for i in range(5):
      x += r.triangular(190,230,210)
    tinas.append(x)
    x=0
  return tinas
```

Generamos 10 simulaciones (al sustituir en n), en dónde ya están sumadas las 50 v.a.´s con su respectivo "bloque".

```python
S(10)
```

**Paso 5:** Para calcular la probabilidad de S > 1000 creamos una función dónde a partir de la restricción de que T sea mayor que 1000 se calculan los eventos favorables (n veces) y se divide en el número de eventos.

```python
def prob(n):
    T = S(n)
    favorables = 0
    for i in range(n):
      if T[i] > 1000:
         favorables += 1
    p = favorables/n
    return p
```

Calculamos la probabilidad cuando n = 10000

```python
prob(10000)
```
