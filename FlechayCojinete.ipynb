Ejemplo de Flecha y Cojinetes


Para este ejemplo vamos a utilizar el método de las 12 uniformes, dónde tenemos $x_1$ con media 1.5 y varianza 0.16, y $x_2$con media 1.48 y varianza 0.09.
Por ello, vamos a tomar $\sigma_1$=0.4 y $\sigma_2$=0.3.



```python
import numpy as np
import random as r
import matplotlib.pyplot as plt
```

```python
def gaussiana(mu,sigma,n):
    muestra = []
    for i in range(n):
        suma = 0
        for j in range(12):
            u = r.random()
            suma = suma + u
        x = mu+sigma*(suma-6)
        muestra.append(x)
    return muestra
```

```python
def probabs(n):
    numero = 0
    x_1 = gaussiana(1.5,0.04,n)
    x_2 = gaussiana(1.48,0.03,n)
    for i in range(n):
        if x_1[i] < x_2[i]:
           numero += 1
    probabs = numero/n
    return probabs
```

```python
probabs(10000)
```

```python
p = probabs(1000)
z = 1.96
error = 0.01

n_nec = (z**2*p*(1-p))/error**2
```
