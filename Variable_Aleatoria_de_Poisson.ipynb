Ejemplo: Simular una variable X tal que $P_1$=0.20, $P_2$=0.15, $P_3$=0.25, $P_4$=0.40 donde $P_j$=P{x=j}.


Usamos la variable de Poisson para contar cuántas veces ocurre un evento en un intervalo.

```python
import numpy as np
import random as r
import matplotlib.pyplot as plt
```

```python
def poissonn(lambdaa,n):
    muestra = []
    for i in range(n):
        R = r.random()
        j = 0
        p = np.exp(-lambdaa)
        F = p
        while R >= F:
            j = j + 1
            p = lambdaa * p / j
            F = F + p
        muestra.append(j)
    return muestra
```

```python
M = poissonn(4,100000)
```

```python
plt.hist(M, bins=100)
```
